## 1. Image Reading and Display

A digital image is stored as a two-dimensional array of pixel intensities. For a grayscale image of size $M \times N$, each pixel value is represented as $I(m,n)$, where $m$ and $n$ are row and column indices. Reading and displaying the image is the first step toward analysis and processing.

```python
# importing necessary libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# reading the image using matplotlib
img = plt.imread('Rose2.png')
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# reading the image using OpenCV
img_cv = cv2.imread('Rose2.png')
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
img_rgb
```

```python
# converting the image to grayscale
img_gry = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img_gry, cmap='gray')
plt.axis('off')
plt.show()
```

```python
img_gry
```

```python
# Extract RGB channels from the loaded image
R = img_rgb [:, :, 0].astype(np.float32)
G = img_rgb [:, :, 1].astype(np.float32)
B = img_rgb [:, :, 2].astype(np.float32)
```

```python
# converting the image to grayscale
# Formula 1 imggray = 0.299*R + 0.587*G + 0.114*B
img_gray_formula1 = (0.299 * R + 0.587 * G + 0.114 * B).astype(np.uint8)

# Method 2: Average method --- IGNORE ---
# imggray = (R + G + B) / 3 --- IGNORE ---
img_gray_formula2 = ((R + G + B) / 3).astype(np.uint8)

fig = plt.figure(figsize=(15, 8))
plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_gray_formula1, cmap='gray')
plt.title('Grayscale Image (Formula 1)')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_gray_formula2, cmap='gray')
plt.title('Grayscale Image (Formula 2)')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img_gray_formula2, cmap='gray')
plt.title('Grayscale Image (Formula 2)')
plt.axis('off')
plt.show()
```
