- 4 adjcency
- 8 adjcency
- d adjcency

```python
# import necessary libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# read a image
img = cv2.imread('cat.png')

# convert the image from BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# show the image
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# convert this image into R, G, B channels
r, g, b = cv2.split(img)
r
```

```python
# convert the image to grayscale
imgray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# show the grayscale image
plt.imshow(imgray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# spilt the cat image into a 5X5 grid

spilt_img = imgray[:5,:5]
spilt_img
```

```python
# 8 - Adjcent pixels function
def adjacent_pixels_8(image, x, y):
    adjacent = []
    for i in range(-1, 2):
        for j in range(-1, 2):
            if (i == 0 and j == 0):
                continue
            adjacent.append(image[x + i, y + j])
    return adjacent
```

```python
# print the adjacent pixels of the pixel at (2, 2)
adjacent_pixels_8(imgray, 2, 2)
```

```python
# 4 - adjacent pixels of the pixel at (2, 2)
def adjacent_pixels_4(image, x, y):
    adjacent = []
    for i in range(-1, 2):
        for j in range(-1, 2):
            if (i == 0 and j == 0) or (i != 0 and j != 0):
                continue
            adjacent.append(image[x + i, y + j])
    return adjacent
```

```python
# print the adjacent pixels of the pixel at (2, 2)
adjacent_pixels_4(imgray, 2, 2)
```

```python
# D - adjacent pixels of the pixel at (2, 2)
def adjacent_pixels_D(image, x, y):
    adjacent = []
    for i in range(-1, 2):
        for j in range(-1, 2):
            if (i == 0 and j == 0) or (i == 0 or j == 0):
                continue
            adjacent.append(image[x + i, y + j])
    return adjacent
```

```python
# print the adjacent pixels of the pixel at (2, 2)
adjacent_pixels_D(imgray, 2, 2)
```
