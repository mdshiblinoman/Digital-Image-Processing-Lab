## Python program for point transformations
**Image negative, log transformations, and power-law transformations.**

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread 
from pathlib import Path
```

```python
# Read and display image
image_path = Path('Rose1.png')
I0 = imread(image_path)
```

```python
# Convert to grayscale and normalize to [0, 1]
import cv2

if I0.ndim == 3:
    I = cv2.cvtColor(I0, cv2.COLOR_RGB2GRAY)
else:
    I = I0

I = I.astype(np.float64) / 255.0
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I0)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(I, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

```python
I_neg = 1.0 - I

plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(I_neg, cmap='gray')
plt.title('Negative Image')
plt.axis('off')
plt.show()
```

```python
# Log transformation function
I_log = np.zeros_like(I)
row, col = I.shape
C = 3
for i in range(row):
    for j in range(col):
        I_log[i, j] = C * np.log1p(I[i, j])
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(I_log, 0, 1), cmap='gray')
plt.title('Log Transformed Image')
plt.axis('off')
plt.show()
```

```python
# Log transformation
I_log2 = 2 * np.log1p(I)
I_log3 = 3 * np.log1p(I)
I_log4 = 4 * np.log1p(I)
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(I_log4, 0, 1), cmap='gray')
plt.title('Log Transformation (factor 4)')
plt.axis('off')
plt.show()
```

```python
# Gamma transformation function
I_gamma = np.zeros_like(I)
C = 0.6
gamma = 0.3
for i in range(row):
    for j in range(col):
        I_gamma[i, j] = C * (I[i, j] ** gamma)
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(I_gamma, 0, 1), cmap='gray')
plt.title('Gamma Transformed Image (factor 3)')
plt.axis('off')
```

```python
# Gamma transformation (gamma greater than 1)
I_gamma4 = I ** 3
I_gamma5 = I ** 4
I_gamma6 = I ** 5
```

```python
fig = plt.figure(figsize=(15, 10))

ax1 = fig.add_subplot(2, 2, 1)
ax1.imshow(I, cmap='gray')
ax1.set_title('Original Image')
ax1.axis('off')

ax2 = fig.add_subplot(2, 2, 2)
ax2.imshow(I_gamma4, cmap='gray')
ax2.set_title('Gamma factor 4')
ax2.axis('off')

ax3 = fig.add_subplot(2, 2, 3)
ax3.imshow(I_gamma5, cmap='gray')
ax3.set_title('Gamma factor 5')
ax3.axis('off')

ax4 = fig.add_subplot(2, 2, 4)
ax4.imshow(I_gamma6, cmap='gray')
ax4.set_title('Gamma factor 6')
ax4.axis('off')

fig.tight_layout()
plt.show()
```

```python
# Gamma transformation (gamma less than 1)
I_gamma1 = I ** 0.6
I_gamma2 = I ** 0.4
I_gamma3 = I ** 0.3
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes[0, 0].imshow(I, cmap='gray')
axes[0, 0].set_title('Original Image')
axes[0, 0].axis('off')
axes[0, 1].imshow(I_gamma1, cmap='gray')
axes[0, 1].set_title('Gamma factor 0.6')
axes[0, 1].axis('off')
axes[1, 0].imshow(I_gamma2, cmap='gray')
axes[1, 0].set_title('Gamma factor 0.4')
axes[1, 0].axis('off')
axes[1, 1].imshow(I_gamma3, cmap='gray')
axes[1, 1].set_title('Gamma factor 0.3')
axes[1, 1].axis('off')
plt.tight_layout()
plt.show()
```
