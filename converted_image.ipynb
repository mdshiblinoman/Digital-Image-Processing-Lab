## Point processing operations: thresholding, intensity adjustments, and contrast stretching. ##

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import exposure
```

## Read and Display Image

```python
# Read image
I0 = cv2.imread('cat.png')
I0_rgb = cv2.cvtColor(I0, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for display
I = cv2.cvtColor(I0, cv2.COLOR_BGR2GRAY)     # Convert to grayscale

# Display both images
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I0_rgb)
axes[0].set_title('Color Image')
axes[0].axis('off')

axes[1].imshow(I, cmap='gray')
axes[1].set_title('Grayscale Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Image Thresholding

```python
# Binary thresholding using Otsu's method
_, Ib = cv2.threshold(I, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Display original and binary
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(Ib, cmap='gray')
axes[1].set_title('Binary (Threshold) Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Intensity Operations

```python
# Adding intensity levels
Id1 = cv2.add(I, 100)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(Id1, cmap='gray')
axes[1].set_title('Intensity Added Image (+100)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Subtracting intensity levels
Id2 = cv2.subtract(I, 50)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(Id2, cmap='gray')
axes[1].set_title('Intensity Subtracted Image (-50)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Multiplying intensity levels
Id3 = cv2.multiply(I, 0.5)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(Id3.astype(np.uint8), cmap='gray')
axes[1].set_title('Intensity Multiplied Image (×0.5)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Dividing intensity levels
Id4 = cv2.divide(I, 0.5)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(np.clip(Id4, 0, 255).astype(np.uint8), cmap='gray')
axes[1].set_title('Intensity Divided Image (÷0.5)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Contrast Stretching

```python
# Add 100 intensity to the original image
I_adjusted = cv2.add(I, 100)

# Apply contrast stretching using scikit-image
# This stretches the histogram based on percentile limits
p2, p98 = np.percentile(I_adjusted, (2, 98))
I_stretched = exposure.rescale_intensity(I_adjusted, in_range=(p2, p98), out_range=(0, 255)).astype(np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(I_adjusted, cmap='gray')
axes[0].set_title('Original Image (Intensity +100)')
axes[0].axis('off')

axes[1].imshow(I_stretched, cmap='gray')
axes[1].set_title('Contrast Stretched Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
