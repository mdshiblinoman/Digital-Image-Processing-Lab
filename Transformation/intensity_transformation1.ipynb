## 1. Intensity Transformation

A point-wise intensity transformation maps each pixel value independently. If the input intensity is $x$ and the output is $y$, then

$$y = T(x)$$

where $T$ is a nonlinear or linear mapping applied to every pixel. This operation is commonly used for brightness adjustment, contrast enhancement, and gamma correction.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread 
from pathlib import Path
```

```python
# Read and display image
image_path = Path('Rose1.png')
I0 = imread(image_path)
```

```python
# Convert to grayscale and normalize to [0, 1]
import cv2

# mannually convert to grayscale
I = 0.299 * I0[:, :, 0] + 0.587 * I0[:, :, 1] + 0.114 * I0[:, :, 2]
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I0)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(I, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

```python
I_neg = 255 - I

plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(I_neg, cmap='gray')
plt.title('Negative Image')
plt.axis('off')
plt.show()
```

```python
# Log transformation function
c = 0.4  # scaling constant   
I_log = c * np.log1p(I)  # log(1 + I)
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(I_log, 0, 1), cmap='gray')
plt.title('Log Transformed Image')
plt.axis('off')
plt.show()
```

```python
# Gamma transformation function
c = 0.01  # scaling constant
I_gamma = c * np.power(I, 0.5)  # gamma = 0.5
```

```python
plt.subplots(1, 2, figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(I_gamma, 0, 1), cmap='gray')
plt.title('Gamma Transformed Image (factor 1)')
plt.axis('off')
```
